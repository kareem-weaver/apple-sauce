```python
# ============================================================
# 0) Setup
# ============================================================
import os
import re
import time
from pathlib import Path
import requests
import pandas as pd
import numpy as np

DATA_DIR = Path("data")
RAW_DIR = DATA_DIR / "raw"
OUT_DIR = DATA_DIR / "processed"
RAW_DIR.mkdir(parents=True, exist_ok=True)
OUT_DIR.mkdir(parents=True, exist_ok=True)

# CENSUS_API_KEY = os.getenv("", "")  # set in env, or paste below
CENSUS_API_KEY = "d508dd731af7f9d40a3952ee423d91a8124c242e"

YEARS = list(range(2017, 2023))  # adjust to match IGS years in your file


# ============================================================
# 1) Load + Clean IGS (handles messy header rows)
# ============================================================
def _find_header_row(df_raw: pd.DataFrame, required_tokens=("Census Tract FIPS", "Year")) -> int:
    """
    Finds the row index that looks like the real header row.
    We look for cells containing key tokens (case-insensitive).
    """
    tokens = [t.lower() for t in required_tokens]
    for i in range(min(len(df_raw), 50)):  # scan first 50 rows
        row = df_raw.iloc[i].astype(str).str.lower().tolist()
        if all(any(tok in cell for cell in row) for tok in tokens):
            return i
    raise ValueError("Could not find header row. Check the raw IGS file formatting.")

def load_igs_any(path: str) -> pd.DataFrame:
    p = Path(path)
    if p.suffix.lower() in [".xlsx", ".xls"]:
        raw = pd.read_excel(p, header=None)
    else:
        raw = pd.read_csv(p, header=None)

    hdr_i = _find_header_row(raw, required_tokens=("Census Tract FIPS", "Year"))
    header = raw.iloc[hdr_i].tolist()
    df = raw.iloc[hdr_i + 1 :].copy()
    df.columns = header
    df = df.dropna(how="all")

    # standardize key columns
    # (keep original names too if you want; this just normalizes)
    col_map = {}
    for c in df.columns:
        if isinstance(c, str):
            c2 = c.strip()
            col_map[c] = c2
    df = df.rename(columns=col_map)

    # Find the likely key columns
    # (your dataset uses these exact names from your column list)
    df["geoid"] = df["Census Tract FIPS code"].astype(str).str.replace(r"\.0$", "", regex=True).str.zfill(11)
    df["year"] = pd.to_numeric(df["Year"], errors="coerce").astype("Int64")

    # Normalize target cols
    # You can change target to pillars later; start with Inclusive Growth Score.
    df["igs_total"] = pd.to_numeric(df.get("Inclusive Growth Score", np.nan), errors="coerce")

    # keep useful IGS columns
    keep = [
        "geoid", "year",
        "igs_total",
        "Place", "Economy", "Community", "Inclusion", "Growth",
        "URBAN CODE", "Is an Opportunity Zone", "BENCHMARK"
    ]
    existing = [c for c in keep if c in df.columns]
    df = df[existing].copy()

    # clean types
    if "URBAN CODE" in df.columns:
        df["urban_code"] = pd.to_numeric(df["URBAN CODE"], errors="coerce")
    if "Is an Opportunity Zone" in df.columns:
        df["is_opp_zone"] = pd.to_numeric(df["Is an Opportunity Zone"], errors="coerce")
    if "BENCHMARK" in df.columns:
        df["benchmark"] = df["BENCHMARK"].astype(str)

    df = df.drop(columns=[c for c in ["URBAN CODE", "Is an Opportunity Zone", "BENCHMARK"] if c in df.columns])

    # Drop rows missing keys
    df = df.dropna(subset=["geoid", "year"])
    df["year"] = df["year"].astype(int)

    return df


# ============================================================
# 2) Census API helpers (ACS 5-year)
# ============================================================
def chunks(lst, n):
    for i in range(0, len(lst), n):
        yield lst[i:i+n]

def get_state_fips(year: int, api_key: str) -> list[str]:
    url = f"https://api.census.gov/data/{year}/acs/acs5"
    params = {"get": "NAME", "for": "state:*", "key": api_key}
    r = requests.get(url, params=params, timeout=60)
    r.raise_for_status()
    data = r.json()
    header, rows = data[0], data[1:]
    df = pd.DataFrame(rows, columns=header)
    return sorted(df["state"].unique().tolist())

def census_get_tracts(year: int, state_fips: str, vars_: list[str], api_key: str) -> pd.DataFrame:
    """
    Pulls ACS 5-year data for ALL tracts in ALL counties of a state.
    Uses the query shape shown in Census API examples:
      ...?get=...&for=tract:*&in=state:06&in=county:*  (wildcards supported)
    """
    base = f"https://api.census.gov/data/{year}/acs/acs5"
    get_str = "NAME," + ",".join(vars_)

    # requests needs duplicate 'in' keys -> use list of tuples
    params = [
        ("get", get_str),
        ("for", "tract:*"),
        ("in", f"state:{state_fips}"),
        ("in", "county:*"),
        ("key", api_key),
    ]

    r = requests.get(base, params=params, timeout=120)
    r.raise_for_status()
    data = r.json()
    header, rows = data[0], data[1:]
    df = pd.DataFrame(rows, columns=header)

    df["year"] = year
    return df

def to_numeric_safe(df: pd.DataFrame, exclude=("NAME", "state", "county", "tract", "year")) -> pd.DataFrame:
    for c in df.columns:
        if c in exclude:
            continue
        df[c] = pd.to_numeric(df[c], errors="coerce")
    return df


# ============================================================
# 3) Define the ACS variables we need (tables/metrics)
# ============================================================

ACS_VARS = [
    # --- population / age (B01001) ---
    "B01001_001E",  # total pop
    "B01001_003E", "B01001_004E", "B01001_005E", "B01001_006E",  # male <18 groups
    "B01001_027E", "B01001_028E", "B01001_029E", "B01001_030E",  # female <18 groups

    # --- race (B02001) ---
    "B02001_001E", "B02001_002E", "B02001_003E", "B02001_005E", "B02001_008E",

    # --- education (B15003) BA+ ---
    "B15003_001E", "B15003_022E", "B15003_023E", "B15003_024E", "B15003_025E",

    # --- employment (B23025) ---
    "B23025_001E", "B23025_002E", "B23025_003E", "B23025_004E", "B23025_005E", "B23025_007E",

    # --- commute time (B08303) ---
    "B08303_001E", "B08303_002E", "B08303_003E", "B08303_004E", "B08303_005E", "B08303_006E", "B08303_007E", "B08303_008E",

    # --- internet subscription (B28002) ---
    "B28002_001E", "B28002_002E",

    # --- rent burden (B25070) ---
    "B25070_001E", "B25070_002E", "B25070_003E", "B25070_004E", "B25070_005E", "B25070_006E",

    # --- owner cost burden (B25091) ---
    "B25091_001E",
    "B25091_003E", "B25091_004E", "B25091_005E", "B25091_006E", "B25091_007E",   # with mortgage <30%
    "B25091_014E", "B25091_015E", "B25091_016E", "B25091_017E", "B25091_018E",   # without mortgage <30%

    # --- occupancy (B25002) ---
    "B25002_001E", "B25002_002E",

    # --- home value (B25077) ---
    "B25077_001E",

    # --- female above poverty (B17001) ---
    "B17001_017E", "B17001_046E",

    # --- gini (B19083) ---
    "B19083_001E",

    # --- per-capita income (B19301) ---
    "B19301_001E",

    # --- early education enrollment proxy (B14003) (3-4 years enrolled public/private, male/female) ---
    "B14003_004E", "B14003_013E", "B14003_032E", "B14003_041E",

    # --- health insurance (B27001) total + uninsured-by-sex/age blocks ---
    "B27001_001E",
    # male uninsured (patterned)
    "B27001_005E","B27001_008E","B27001_011E","B27001_014E","B27001_017E","B27001_020E","B27001_023E","B27001_026E","B27001_029E",
    # female uninsured (patterned)
    "B27001_033E","B27001_036E","B27001_039E","B27001_042E","B27001_045E","B27001_048E","B27001_051E","B27001_054E","B27001_057E",
]


# ============================================================
# 4) Download ACS for all states + years (cached)
# ============================================================
def download_acs_all(years: list[int], api_key: str, max_vars_per_call: int = 45) -> pd.DataFrame:
    if not api_key:
        raise ValueError("Missing CENSUS_API_KEY. Set env var CENSUS_API_KEY or paste it into the notebook.")

    all_parts = []
    for year in years:
        year_out = RAW_DIR / f"acs5_tract_{year}.parquet"
        if year_out.exists():
            print(f"[cache] {year_out}")
            all_parts.append(pd.read_parquet(year_out))
            continue

        states = get_state_fips(year, api_key)
        print(f"[year {year}] states={len(states)}")

        state_frames = []
        for st in states:
            merged = None
            for var_chunk in chunks(ACS_VARS, max_vars_per_call):
                df_chunk = census_get_tracts(year, st, var_chunk, api_key)
                df_chunk = to_numeric_safe(df_chunk)

                keys = ["state", "county", "tract", "NAME", "year"]
                if merged is None:
                    merged = df_chunk
                else:
                    merged = merged.merge(df_chunk, on=keys, how="outer")

                time.sleep(0.25)  # be polite to the API

            state_frames.append(merged)

        df_year = pd.concat(state_frames, ignore_index=True)
        df_year.to_parquet(year_out, index=False)
        all_parts.append(df_year)

    return pd.concat(all_parts, ignore_index=True)


# ============================================================
# 5) Feature engineering to mirror IGS-style public proxies
# ============================================================
def safe_div(n, d):
    return np.where((d is None) | (pd.isna(d)) | (d == 0), np.nan, n / d)

def make_features(acs: pd.DataFrame) -> pd.DataFrame:
    # GEOID
    acs["state"] = acs["state"].astype(str).str.zfill(2)
    acs["county"] = acs["county"].astype(str).str.zfill(3)
    acs["tract"] = acs["tract"].astype(str).str.zfill(6)
    acs["geoid"] = acs["state"] + acs["county"] + acs["tract"]

    out = pd.DataFrame({"geoid": acs["geoid"], "year": acs["year"]})

    # population / age
    out["pop_total"] = acs["B01001_001E"]
    out["pop_under5"] = acs["B01001_003E"] + acs["B01001_027E"]
    out["pop_under18"] = (
        acs["B01001_003E"] + acs["B01001_004E"] + acs["B01001_005E"] + acs["B01001_006E"] +
        acs["B01001_027E"] + acs["B01001_028E"] + acs["B01001_029E"] + acs["B01001_030E"]
    )
    out["share_under18"] = safe_div(out["pop_under18"], out["pop_total"])

    # race shares
    out["share_white"] = safe_div(acs["B02001_002E"], acs["B02001_001E"])
    out["share_black"] = safe_div(acs["B02001_003E"], acs["B02001_001E"])
    out["share_asian"] = safe_div(acs["B02001_005E"], acs["B02001_001E"])
    out["share_two_plus"] = safe_div(acs["B02001_008E"], acs["B02001_001E"])

    # education BA+
    ba_plus = acs["B15003_022E"] + acs["B15003_023E"] + acs["B15003_024E"] + acs["B15003_025E"]
    out["ba_plus_share_25p"] = safe_div(ba_plus, acs["B15003_001E"])

    # employment
    out["lfpr_16p"] = safe_div(acs["B23025_002E"], acs["B23025_001E"])
    out["unemp_rate"] = safe_div(acs["B23025_005E"], acs["B23025_003E"])

    # commute under 35 minutes
    commute_under35 = (
        acs["B08303_002E"] + acs["B08303_003E"] + acs["B08303_004E"] + acs["B08303_005E"] +
        acs["B08303_006E"] + acs["B08303_007E"] + acs["B08303_008E"]
    )
    out["commute_under35_share"] = safe_div(commute_under35, acs["B08303_001E"])

    # internet subscription
    out["internet_sub_share"] = safe_div(acs["B28002_002E"], acs["B28002_001E"])

    # affordable housing proxy (cost burden <30%)
    rent_affordable = (
        acs["B25070_002E"] + acs["B25070_003E"] + acs["B25070_004E"] + acs["B25070_005E"] + acs["B25070_006E"]
    )
    owner_affordable = (
        (acs["B25091_003E"] + acs["B25091_004E"] + acs["B25091_005E"] + acs["B25091_006E"] + acs["B25091_007E"]) +
        (acs["B25091_014E"] + acs["B25091_015E"] + acs["B25091_016E"] + acs["B25091_017E"] + acs["B25091_018E"])
    )
    denom_hh = acs["B25070_001E"] + acs["B25091_001E"]
    out["affordable_housing_share"] = safe_div(rent_affordable + owner_affordable, denom_hh)

    # female above poverty
    out["female_above_poverty_share"] = safe_div(acs["B17001_046E"], (acs["B17001_017E"] + acs["B17001_046E"]))

    # early education enrollment proxy (3-4 enrolled / under5 pop)
    enrolled_3_4 = acs["B14003_004E"] + acs["B14003_013E"] + acs["B14003_032E"] + acs["B14003_041E"]
    out["early_ed_enroll_share"] = safe_div(enrolled_3_4, out["pop_under5"])

    # health insurance coverage (1 - uninsured/total)
    uninsured = (
        acs["B27001_005E"] + acs["B27001_008E"] + acs["B27001_011E"] + acs["B27001_014E"] + acs["B27001_017E"] +
        acs["B27001_020E"] + acs["B27001_023E"] + acs["B27001_026E"] + acs["B27001_029E"] +
        acs["B27001_033E"] + acs["B27001_036E"] + acs["B27001_039E"] + acs["B27001_042E"] + acs["B27001_045E"] +
        acs["B27001_048E"] + acs["B27001_051E"] + acs["B27001_054E"] + acs["B27001_057E"]
    )
    out["insured_share"] = 1.0 - safe_div(uninsured, acs["B27001_001E"])

    # gini + income + housing occupancy/value
    out["gini"] = acs["B19083_001E"]
    out["per_capita_income"] = acs["B19301_001E"]
    out["occupied_units"] = acs["B25002_002E"]
    out["median_home_value"] = acs["B25077_001E"]

    # growth features (year-over-year within tract)
    out = out.sort_values(["geoid", "year"])
    for col in ["per_capita_income", "occupied_units", "median_home_value"]:
        out[f"{col}_growth"] = out.groupby("geoid")[col].pct_change()

    return out


# ============================================================
# 6) Merge ACS features with IGS
# ============================================================
def merge_igs_acs(igs: pd.DataFrame, feats: pd.DataFrame) -> pd.DataFrame:
    merged = igs.merge(feats, on=["geoid", "year"], how="left")
    return merged


# ============================================================
# 7) Run pipeline
# ============================================================
# 7a) Load IGS

igs_path = r"C:\Users\jabba\Desktop\Code\machine_learning\AUC_mastercard_challenge\src\Inclusive_Growth_Score_Data.csv"

igs = load_igs_any(igs_path)
igs.to_parquet(OUT_DIR / "igs_clean.parquet", index=False)

# 7b) Download ACS for needed years
acs = download_acs_all(sorted(igs["year"].unique().tolist()), CENSUS_API_KEY)
acs.to_parquet(OUT_DIR / "acs_raw.parquet", index=False)

# 7c) Feature engineering
feats = make_features(acs)
feats.to_parquet(OUT_DIR / "acs_features.parquet", index=False)

# 7d) Merge
model_df = merge_igs_acs(igs, feats)
model_df.to_parquet(OUT_DIR / "igs_x_acs.parquet", index=False)

print(model_df.shape)
print(model_df[["igs_total","affordable_housing_share","internet_sub_share","insured_share"]].describe())
```
